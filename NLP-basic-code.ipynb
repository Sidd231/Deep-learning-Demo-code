```python
df = pd.read_csv("/kaggle/input/datasets/praveengovi/emotions-dataset-for-nlp/test.txt" , sep = ';' ,header = None, names = ["text" , "emotion"])
```

```python
df.isnull().sum()
```

```python
uni_emotion = df["emotion"].unique()
uni_num = {}
for i in range(len(uni_emotion)):
    uni_num[uni_emotion[i]] = i
```

```python
uni_num
```

```python
df['emotion'] = df["emotion"].map(uni_num)
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))

df['text'] = df['text'].apply(remove_punc)
```

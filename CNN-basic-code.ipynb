```python
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.model_selection import train_test_split

from sklearn.linear_model import Perceptron

from sklearn.metrics import accuracy_score, classification_report,confusion_matrix

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Dense 
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import Flatten

from tensorflow.keras.layers import MaxPooling2D
from tensorflow.keras.layers import Dropout

from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv("/kaggle/input/datasets/zalando-research/fashionmnist/fashion-mnist_train.csv")
df_test = pd.read_csv("/kaggle/input/datasets/zalando-research/fashionmnist/fashion-mnist_test.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X_train = df.drop("label" , axis = 1).values
y_train = df["label"].values
X_test = df_test.drop("label" , axis = 1).values
y_test = df_test["label"].values
```

```python
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
# doing this so that model train better 
```

```python
X_train_img = X_train.reshape(-1,28,28)
X_test_img = X_test.reshape(-1,28,28)
```

```python
X_train_img
```

```python
y_train_cat = to_categorical(y_train,10)
y_test_cat = to_categorical(y_test, 10)
```

```python
perceptron = Sequential([
    Flatten(input_shape = (28,28)),
    Dense(10 , activation = "softmax")
])
```

```python
perceptron.compile(optimizer = "sgd" , loss = "categorical_crossentropy" , metrics = ["accuracy"])
```

```python
history_percp = perceptron.fit(X_train_img, y_train_cat, epochs = 10,
batch_size=32 , validation_data=(X_test_img, y_test_cat), verbose=1)
```

```python
acc = perceptron.evaluate(X_test_img, y_test_cat, verbose = 0)[1]
acc 
```

```python
ann = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])
```

```python
ann.compile(optimizer = "adam", loss="categorical_crossentropy" , metrics = ["accuracy"])
```

```python
history_ann = ann.fit(X_train_img, y_train_cat, epochs= 10,batch_size = 32,
                     validation_data = (X_test_img, y_test_cat), verbose= 1) 
```

```python
acc = ann.evaluate(X_test_img , y_test_cat, verbose = 0)[1]
```
